# Assignment_2_KMeans_Clustering

Apply clustering techniques using unsupervised learning.


## Aim
Apply K-Means clustering as an unsupervised learning technique.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RESULT_DIR = "/content/results"
os.makedirs(RESULT_DIR, exist_ok=True)

print("Result folder:", RESULT_DIR)

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

iris = load_iris()
X = StandardScaler().fit_transform(iris.data)

model = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = model.fit_predict(X)

summary = pd.DataFrame({
    "Cluster":[0,1,2],
    "Number_of_samples":[int((labels==0).sum()),int((labels==1).sum()),int((labels==2).sum())]
})
display(summary)

plt.figure(figsize=(7,5))
plt.scatter(X[:,0], X[:,1], c=labels, cmap="viridis", s=45)
plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")
plt.title("K-Means Clustering")
plt.tight_layout()
plt.savefig(f"{RESULT_DIR}/kmeans_clusters.png", dpi=150)
plt.show()

summary.to_csv(f"{RESULT_DIR}/cluster_summary.csv", index=False)
pd.DataFrame({"sample":np.arange(len(labels)), "cluster":labels}).to_csv(
    f"{RESULT_DIR}/cluster_assignments.csv", index=False)
```
